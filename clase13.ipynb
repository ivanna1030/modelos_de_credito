**The Misleading Nature of Correlations:**

- **Simpson's Paradox:** A trend appears in different groups of data but disappears or reverses when these groups are combined.

- **Spurious Correlations:** Two variables can be correlated even when there is no causal relationship between them.

- **Nonlinear Relationships:** The relationship between variables is not linear, so a linear correlation misses it.

- **Causation vs Correlation:** Correlation does not imply causation. The causal mechanism must be examined.


**Common Biases in Financial Analysis:**

- **Confirmation bias:** The tendency to search for, interpret, favor, and recall information in a way that confirms or supports one's prior beliefs or values.

- **Anchoring bias:** The tendency to rely too heavily, or "anchor," on one trait or piece of information when making decisions.

- **Availability bias:** The tendency to overestimate the likelihood of events with greater "availability" in memory, which can be influenced by how recent or emotionally charged the memories are.

- **Overconfidence:** The tendency to overestimate one's own abilities, performance, level of control, or odds of success.

**Strategies to Mitigate Analytical Biases:**

- **Consider the opposite:** Think of reasons why your initial judgement might be wrong.

- **Get an outside perspective:** Consult unbiased third parties to get a different viewpoint.

- **Avoid anchoring:** Don't rely too heavily on the first piece of information.

- **Watch for conformity bias:** Resist the urge to align with the majority opinion.

- **Challenge assumptions:** Question if your assumptions are based on facts.
