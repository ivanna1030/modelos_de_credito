# Guía para crear un modelo de crédito o un modelo de negocio basado en el crédito

**Definición del objetivo:** Definir claramente estos elementos fundamentales sentará las bases para desarrollar un modelo de crédito efectivo.

- **Definir el tipo de crédito:** Especificar si será crédito personal, hipotecario, automotriz, pyme, etc.

- **Identificar el público objetivo:** Analizar las necesidades y perfil de los posibles clientes.

- **Establecer los productos y servicios:** Determinar las líneas y montos de crédito, plazos, tasas de interés, etc.

- **Definir políticas y procedimientos:** Incluir requisitos, evaluación de riesgos, aprobación de créditos, cobranza, etc.

**Recopilación de datos:**

- Realizar encuestas, entrevistas y análisis de datos permiten obtener información valiosa sobre los clientes potenciales. Las pruebas piloto ayudan a probar el modelo antes de lanzarlo.

**Investigación:**

- INVESTIGACIÓN, LEER, VIDEOS, PREGUNTAR, MYSTERY SHOPPING.

**Desarrollo del modelo:**

- **Definir el método:** Seleccionar el método más adecuado para desarrollar el modelo, como investigación, análisis estadístico, aprendizaje automático o inteligencia artificial.

- **Utilizar los datos recopilados:** Usar los datos recopilados sobre los clientes y el mercado para entrenar y probar el modelo.

- **Ajustar a los objetivos:** Asegurarse de que el modelo desarrollado se ajusta al objetivo definido inicialmente.

- **Validar con políticas existentes:** Validar que el modelo desarrollado sigue las políticas y procedimientos establecidos.

**MONITOREO Y MEJORAS:**

- Establecer KPIs y seguimiento, se puede hacer en herramientas de dashboard.
